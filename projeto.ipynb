```python

from pathlib import Path
import shutil

# Arquivo original
arquivo_original = Path("hotel_bookings.csv")

# Pasta destinada à modelagem
pasta_modelagem = Path("dados_modelagem")
pasta_modelagem.mkdir(exist_ok=True)

# Caminho da cópia
arquivo_copia = pasta_modelagem / "hotel_bookings_modelagem.csv"

# Criar a cópia sem sobrescrever uma cópia existente
if arquivo_original.exists():
    if not arquivo_copia.exists():
        shutil.copy2(arquivo_original, arquivo_copia)
        print("Cópia criada com sucesso!")
    else:
        print("A cópia já existe. Ela foi preservada.")
else:
    print("Arquivo original não encontrado. Verifique o caminho.")

print("Original:", arquivo_original.resolve())
print("Cópia:", arquivo_copia.resolve())
```

```python
# Criar faixas de antecedência da reserva
faixas = [0, 7, 30, 60, 90, 180, 365, float("inf")]

nomes_faixas = [
    "0 a 7 dias",
    "8 a 30 dias",
    "31 a 60 dias",
    "61 a 90 dias",
    "91 a 180 dias",
    "181 a 365 dias",
    "Mais de 365 dias"
]

df_modelagem["faixa_lead_time"] = pd.cut(
    df_modelagem["lead_time"],
    bins=faixas,
    labels=nomes_faixas,
    include_lowest=True
)

# Calcular a quantidade de reservas em cada faixa
quantidade = df_modelagem.groupby(
    "faixa_lead_time",
    observed=False
)["is_canceled"].count()

# Calcular a porcentagem de cancelamentos em cada faixa
cancelamentos = df_modelagem.groupby(
    "faixa_lead_time",
    observed=False
)["is_canceled"].mean() * 100

# Montar tabela de resultados
analise_lead_time = pd.DataFrame({
    "Quantidade de reservas": quantidade,
    "Cancelamentos (%)": cancelamentos
})

display(analise_lead_time.round(2))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(
    analise_lead_time.index.astype(str),
    analise_lead_time["Cancelamentos (%)"]
)

plt.title("Taxa de cancelamento por antecedência da reserva")
plt.xlabel("Antecedência da reserva")
plt.ylabel("Cancelamentos (%)")

plt.xticks(rotation=30, ha="right")
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

```python
# Quantidade de reservas por tipo de depósito e situação
reservas_deposito = pd.crosstab(
    df_modelagem["deposit_type"],
    df_modelagem["is_canceled"]
)

# Percentual de cancelamentos por tipo de depósito
percentual_deposito = pd.crosstab(
    df_modelagem["deposit_type"],
    df_modelagem["is_canceled"],
    normalize="index"
) * 100

percentual_deposito.columns = [
    "Não canceladas (%)",
    "Canceladas (%)"
]

print("Quantidade de reservas:")
display(reservas_deposito)

print("Percentual de cancelamentos:")
display(percentual_deposito.round(2))
```
