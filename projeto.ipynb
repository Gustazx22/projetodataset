## Importa Biblioteca
Esta célula importa a biblioteca **Pandas** e lê o arquivo `hotel_bookings.csv`, criando o DataFrame `df` com todos os dados da planilha.

```python
import pandas as pd

df = pd.read_csv("hotel_bookings.csv")
```

## Dados dos por pais
Esta célula agrupa os dados por país e calcula a média das colunas numéricas para cada país.

```python
df.groupby("country").mean(numeric_only=True).round(2)
```

## Conta a Frequencia
Esta célula conta a frequência de cada valor da coluna analisada (como país, agente ou empresa).

```python
df["country"].value_counts(dropna=False)
```

## Calculo das Reservas
Esta célula calcula a porcentagem de reservas por país (`country`).

```python
df["country"].value_counts(normalize=True, dropna=False).mul(100).round(2)
```
